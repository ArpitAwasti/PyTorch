# ***Pytorch Fundamentals***

**Simple Array manipulation/creation**

```python
import torch
import numpy

# create tensor from numpy array
numpy_tensor = numpy.random.randn(2, 2)
numpy_tensor
```

```python
# create tensor from pytorch tensor
pytorch_tensor = torch.randn(2, 2)
pytorch_tensor
```

```python
# convert numpy array to pytorch array
torch.Tensor(numpy_tensor)
```

```python
# or another way
torch.from_numpy(numpy_tensor)
```

```python
# convert torch tensor to numpy representation
pytorch_tensor.numpy()
```

# ***Variable and Gradients***

**Variable Creation**

```python
from torch.autograd import Variable

# create variable
a = Variable(torch.ones((2, 3)), requires_grad=True)

# access variable tensor
a.data
```

```python
# access variable gradient
a.grad
```

**Compute Gradient**

```python
x = Variable(torch.ones(2), requires_grad=True)
y = 5 * (x + 2) ** 2
```

```python
# backward should be called only on a scalar
o = (1 / 2) * torch.sum(y)
```

```python
# compute backward
o.backward()
```

```python
# now we have the gradients of x
x.grad
```
